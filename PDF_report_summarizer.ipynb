```python
import fitz
import google.generativeai as genai
import re
import json
import os
from google.colab import files

# Set Gemini API key securely
os.environ["GEMINI_API_KEY"] = "your_gemini_api_key"
genai.configure(api_key=os.environ["GEMINI_API_KEY"])

# Function to extract text from a PDF file
def extract_text_from_pdf(pdf_path):
    doc = fitz.open(pdf_path)
    text = "\n".join([page.get_text("text") for page in doc])
    return text

# Function to clean and preprocess extracted text
def clean_text(text):
    text = re.sub(r'\s+', ' ', text).strip()
    return text

# Function to summarize key financial insights using Gemini API
def summarize_text(text):
    model = genai.GenerativeModel("gemini-pro")
    prompt = f"""
    Extract key investor insights from the following financial document:
    - Future Growth Prospects
    - Key Business Changes
    - Financial Triggers Impacting Earnings
    - Risks & Opportunities

    Document:
    """
    response = model.generate_content(prompt + text)
    return response.text.strip()

# Function to upload PDF file in Google Colab
def upload_pdf():
    uploaded = files.upload()
    for filename in uploaded.keys():
        return filename

# Main execution flow
def process_pdf(pdf_path):
    raw_text = extract_text_from_pdf(pdf_path)
    cleaned_text = clean_text(raw_text)
    summary = summarize_text(cleaned_text)
    return summary

# Upload file in Colab
pdf_path = upload_pdf()
summary = process_pdf(pdf_path)
print("\n### Key Insights:\n", summary)
```
